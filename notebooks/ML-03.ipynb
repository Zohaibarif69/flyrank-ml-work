## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I would frame my lane as a **classification** task. The goal is to classify each customer support ticket into an urgency level, such as **urgent** or **not urgent**. Classification fits because the output is a discrete label rather than a continuous number.


```python
# Check the number of classes we plan to predict
classes = ["urgent", "not_urgent"]
print("Number of classes:", len(classes))
print("Classes:", classes)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is whether a support ticket is **urgent**. The label can come from an observed outcome, such as whether the ticket received an emergency escalation or was resolved within a short priority window. If that outcome is not available, a clearly defined operational rule can be used as a proxy. The proxy should be treated as a decision-support label, not as a perfect measure of urgency.


```python
# Example target definition for a small demonstration dataset
tickets = [
    {"ticket_id": 1, "urgent": 1},
    {"ticket_id": 2, "urgent": 0},
    {"ticket_id": 3, "urgent": 1},
    {"ticket_id": 4, "urgent": 0},
    {"ticket_id": 5, "urgent": 0},
    {"ticket_id": 6, "urgent": 1},
]

urgent_count = sum(ticket["urgent"] for ticket in tickets)
total_count = len(tickets)

print("Total tickets:", total_count)
print("Urgent tickets:", urgent_count)
print("Urgent share:", round(urgent_count / total_count, 2))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I would use **F1 score** as the main success metric. F1 balances precision and recall, which matters because missing an urgent ticket and incorrectly flagging a non-urgent ticket are both useful errors to measure. As a starting point, an **F1 score of 0.80 or higher** would be considered a useful result, but the threshold should be validated against real operational outcomes.


```python
# Example success threshold
target_f1 = 0.80

print("Target F1 score:", target_f1)
print("A model reaching or exceeding this is a candidate for further evaluation.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could flag a ticket whenever it contains a word such as “outage,” but urgency can depend on several signals at the same time. Message content, previous replies, length, and other measured features may interact in ways that are difficult to capture with one if-statement. ML can learn these patterns from observed examples and provide **decision-support** rather than relying on one fixed rule.


```python
# Show that multiple measured features are available for the task
feature_columns = [
    "message_length",
    "contains_outage_word",
    "previous_replies"
]

print("Features used for decision-support:")
for column in feature_columns:
    print("-", column)

print("\nNumber of features:", len(feature_columns))
```
