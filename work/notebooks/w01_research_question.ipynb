## 3. Quick Look at the Data

The starter dataset contains **30,000** content pages from **32** anonymized clients. On average, each page received approximately **5,200.37 impressions** over the last 90 days, with an average **CTR of 0.51** and an average search position of **16.34**. These numbers show meaningful variation in search performance across pages, making the dataset suitable for prioritizing which content should be reviewed first. This supports the Refresh / Content Opportunity Scoring lane because it focuses on identifying pages that may benefit from review based on measurable search and engagement signals.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Basic statistics
print("Number of rows:", len(df))
print("Unique content pages:", df["content_id"].nunique())
print("Unique clients:", df["client_id"].nunique())
print("Average 90-day impressions:", round(df["impressions_90d"].mean(), 2))
print("Average CTR:", round(df["ctr"].mean(), 2))
print("Average Position:", round(df["avg_position"].mean(), 2))
```
