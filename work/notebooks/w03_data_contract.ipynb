## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

One row is one **client × content × report date** observation. I will use **March 2026** as the development window and keep later observations as future outcome data. The final June 2026 `_sample` is excluded from development because it represents the final outcome month rather than a random sample.


```python
# ML-04 — Section 1: verify the unit of analysis and time window

import duckdb
from huggingface_hub import hf_hub_download

march_file = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="fact_content_daily_performance/month=2026-03/data_0.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

march_path = march_file.replace("\\", "/")

grain_check = duckdb.sql(f"""
SELECT
    COUNT(*) AS row_count,
    COUNT(DISTINCT
        CAST(report_date AS VARCHAR) || '|' ||
        client_hash_id || '|' ||
        content_hash_id
    ) AS distinct_client_content_date
FROM read_parquet('{march_path}')
""").df()

date_check = duckdb.sql(f"""
SELECT
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date,
    COUNT(DISTINCT report_date) AS distinct_dates
FROM read_parquet('{march_path}')
""").df()

print("Grain check:")
display(grain_check)

print("Date-window check:")
display(date_check)
```

## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

**Features:** `gsc_impressions`, `gsc_clicks`, `gsc_avg_position`, `ga4_sessions`, `ga4_engaged_sessions`, and AI referral sessions can describe performance known at the decision moment. I will keep the final model feature set to five fields.

**Label:** the future outcome will be whether the same client-content pair receives AI referral sessions in the following month. This is a defined proxy for future AI-search visibility, not a causal measure.

**Context:** `report_date`, `client_hash_id`, and `content_hash_id` identify when and which client-content observation is being measured.

**Excluded:** future-month outcome fields and label-derived fields are excluded from the honest feature set. I also exclude the June `_sample` table from development because it is the final month and is not a random sample. These exclusions reduce leakage and keep the features available at the decision moment.


## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

I will verify three things on March 2026 data: the row grain, the size and date span of the slice, and the availability of the required search data. Availability is measured with `IS TRUE` rather than assuming that missing values mean unavailable.


```python
# Query 1 — Grain

display(duckdb.sql(f"""
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT
        CAST(report_date AS VARCHAR) || '|' ||
        client_hash_id || '|' ||
        content_hash_id
    ) AS distinct_client_content_date,
    COUNT(*) =
        COUNT(DISTINCT
            CAST(report_date AS VARCHAR) || '|' ||
            client_hash_id || '|' ||
            content_hash_id
        ) AS grain_is_unique
FROM read_parquet('{march_path}')
""").df())
```

```python
# Query 2 — March row count and date span

display(duckdb.sql(f"""
SELECT
    COUNT(*) AS row_count,
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date,
    COUNT(DISTINCT report_date) AS number_of_dates
FROM read_parquet('{march_path}')
""").df())
```

```python
# Query 3 — GSC availability

display(duckdb.sql(f"""
SELECT
    COUNT(*) AS total_rows,
    COUNTIF(gsc_data_available IS TRUE) AS gsc_available_rows,
    ROUND(
        100.0 * COUNTIF(gsc_data_available IS TRUE) / COUNT(*),
        2
    ) AS gsc_available_pct
FROM read_parquet('{march_path}')
""").df())
```

### Five initial features

1. **`gsc_impressions`** — available when? Because Search Console impressions are measured for the current observation period before the future outcome is evaluated.

2. **`gsc_clicks`** — available when? Because clicks are observed during the development period and do not require knowing the following month's outcome.

3. **`gsc_avg_position`** — available when? Because average search position is measured from the current-period Search Console data.

4. **`ga4_sessions`** — available when? Because sessions are observed during the current period when the decision is made.

5. **`sessions_ai`** — available when? Because AI referral sessions are measured in the current period and can describe existing AI-search visibility before the future outcome window.

These are decision-time measurements. Future-month values are not used as features.


```python
# Five-feature frame for March 2026

features = duckdb.sql(f"""
SELECT
    report_date,
    client_hash_id,
    content_hash_id,
    gsc_impressions,
    gsc_clicks,
    gsc_avg_position,
    ga4_sessions,
    sessions_ai
FROM read_parquet('{march_path}')
WHERE report_date >= DATE '2026-03-01'
  AND report_date <= DATE '2026-03-31'
""").df()

print("Feature frame shape:", features.shape)
display(features.head())
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

The data cannot show causation between search activity and later performance. Coverage is also unbalanced because GSC and GA4 availability can differ across clients and dates. Some early observations may be GSC-only, so comparisons across sources need care. Daily observations for the same client and content can overlap across time windows, so they should not automatically be treated as independent. The March window is useful for development, but the future outcome window must remain separate from the features. These limits make the analysis directional and suitable for decision-support rather than a causal claim.
